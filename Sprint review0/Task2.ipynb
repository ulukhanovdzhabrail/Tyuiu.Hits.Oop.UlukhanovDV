```polyglot-notebook
using System.Collections.Generic;
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;


    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity += amount;
        Console.WriteLine($"Количество товара '{name}' изменено на {amount}. Теперь: {quantity}");
    }

    public string GetProductInfo()
    {
        return $"Название: {name}, Цена: {price} ₽, Количество: {quantity}, Категория: {category}, Описание: {description}";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;

    private List<Product> products = new List<Product>();

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
        Console.WriteLine($"Продавец {name} добавил товар: {product.Name}");
    }

    public void SellProduct(Product product, int quantity)
    {
        product.UpdateQuantity(-quantity);
        Console.WriteLine($"Продавец {name} продал {quantity} шт. товара '{product.Name}'");
    }

    public string GetSellerInfo()
    {
        return $"Имя: {name}, ID сотрудника: {employeeId}, Зарплата: {salary} ₽, Контакты: {contactInfo}";
    }
}

public class Store
{

    private string storeName;
    private string location;
    public string storeHours;

    private List<Seller> sellers = new List<Seller>();
    private List<Product> products = new List<Product>();

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
        Console.WriteLine($"В магазин '{storeName}' добавлен продавец: {seller.Name}");
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
    }

    public void ListProducts()
    {
        Console.WriteLine($"Товары магазина '{storeName}':");
        foreach (Product p in products)
        {
            Console.WriteLine(" - " + p.GetProductInfo());
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {storeName}, Адрес: {location}, Часы работы: {storeHours}, Продавцов: {sellers.Count}, Товаров: {products.Count}";
    }
}
Product bread = new Product("Хлеб", 50.0m, 20, "Еда", "Свежий ржаной");
Seller Ivanovich = new Seller("Иван", "EMP-001", 50000.0m, "+7-888-888-88-88");
Store shop = new Store("Пятёрочка", "Тюмень ул. Мельникайте", "08:00 - 23:00");
Console.WriteLine(bread.GetProductInfo());
Console.WriteLine($"Общая стоимость {bread.GetTotalPrice()}");
bread.UpdateQuantity(10);
Console.WriteLine($"Общая стоимость {bread.GetTotalPrice()}");
Console.WriteLine();
Console.WriteLine(Ivanovich.GetSellerInfo());
Ivanovich.AddProduct(bread);
Ivanovich.SellProduct(bread, 5);
Console.WriteLine();
Console.WriteLine(shop.GetStoreInfo());
shop.AddSeller(Ivanovich);
shop.AddProduct(bread);
shop.ListProducts();
Console.WriteLine(shop.GetStoreInfo());



```
