```polyglot-notebook
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветёт.");
    }

    public void Grow(float amount)
    {
        Height += amount;
        Console.WriteLine($"{Name} выросло на {amount} см. Текущая высота: {Height} см.");
    }
}

public class Flower : Plant 
{
    public string Colour { get; set; }
    public int PetalCount { get; set; }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }
    public void Wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать...");
    }
}

public class Tree : Plant 
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void AgeUp(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} выросло на {years} лет. Новый возраст: {Age} лет.");
    }
}

Flower rose = new Flower();
rose.Name = "Роза";
rose.Type = "Цветок";
rose.Height = 25f;
rose.Colour = "Красный";
rose.PetalCount = 32;
rose.Bloom();
rose.Grow(5);
rose.Wither();
Tree oak = new Tree();
oak.Name = "Дуб";
oak.Type = "Дерево";
oak.Height = 500f;
oak.Age = 50;
oak.LeafType = "Лиственные";
oak.Bloom();
oak.Grow(20);
oak.AgeUp(5);


```
