Дан целочисленный массив numbers со следующими элементами { 5, 8, 1, 3, 9, 6, 7, 2 }, нужно выполнить следующие операции:

1. Найти минимальное и максимальное значения в массиве.
2. Найти сумму всех элементов массива.
3. Найти среднее арифметическое всех элементов массива.
4. Отсортировать массив в порядке возрастания.
5. Проверить, содержит ли массив определенное число.

```polyglot-notebook
int [] numbers = {5, 8, 1, 3, 9, 6, 7, 2};

int min = numbers.Min();
int max = numbers.Max();

Console.WriteLine($"Минимальное значение в массиве: {min}");
Console.WriteLine($"Максимальное значение в массиве: {max}");
```

```polyglot-notebook
int summa = numbers.Sum();

Console.WriteLine($"Сумма всех элементов массива: {summa}")
```

```polyglot-notebook
double average = numbers.Average();

Console.WriteLine($"Среднее арифметическое всех элементов массива: {average}");
```

```polyglot-notebook
int [] numSort = numbers.ToArray();
Array.Sort(numSort);

Console.WriteLine($"Отсортированный массив в порядке возрастания: {string.Join( ", ", numSort)}");
```

```polyglot-notebook
int numFind = 0;
bool contains = numbers.Contains(numFind);
if (contains == true)
{
    Console.WriteLine($"Массив содержит число {numFind}");
}
else
{
    Console.WriteLine($"Массив не содержит число {numFind}");
}
```
