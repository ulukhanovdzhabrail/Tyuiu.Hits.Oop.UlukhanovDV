```
using System;
using System.Collections.Generic;

public class Review
{
    //Атрибуты

    public int ReviewId { get; set; }
    public string Text { get; set; }

    private int _rating;
    public int Rating
    {
        get { return _rating; }
        set
        {
            if (value < 1 || value > 5)
                throw new ArgumentException("Рейтинг должен быть от 1 до 5");
            _rating = value;
        }
    }

    //Конструктор

    public Review(int reviewId, string text, int rating)
    {
        ReviewId = reviewId;
        Text = text;
        Rating = rating;    
    }

    //Методы

    public virtual void DisplayReview()
    {
        Console.WriteLine($"[Отзыв #{ReviewId}] {Text} (рейтинг: {Rating})");
    }

    public virtual void RateProduct(int rating)
    {
        Rating = rating;
        Console.WriteLine($"Рейтинг отзыва #{ReviewId} установлен: {rating}");
    }

    public virtual string GetReviewDetails()
    {
        return $"ID: {ReviewId}, Текст: {Text}, Рейтинг: {Rating}";
    }
}

public class ServiceReview : Review
{
    public DateTime VisitDate { get; set; }

    //Конструктор для ServiceReview

    public ServiceReview(int reviewId, string text, int rating, DateTime visitDate) : base(reviewId, text, rating)
    {
       VisitDate = visitDate;
    }
    public override void DisplayReview()
    {
        base.DisplayReview();
        Console.WriteLine($"Дата посещения: {VisitDate:dd.MM.yyyy}");
    }

}
public class ProductReview : Review 
{
    public string ProductId { get; set; }

    //Конструктор для ProductReview

    public ProductReview(int reviewId, string text, int rating, string productId) : base(reviewId, text, rating) 
    {
        ProductId = productId;
    }

    public override void RateProduct(int rating)
    {
        base.RateProduct(rating);
        Console.WriteLine($"Рейтинг {rating} привязан к товару {ProductId}");
    }
}

        var vr = new ProductReview(1, "Отличный товар", 5, "RU-21212");
        vr.RateProduct(5);
        vr.DisplayReview();
        var pr = new Review(2, "Отличный товар", 5);
        pr.DisplayReview();
        var rr = new ServiceReview(3, "Good", 4, new DateTime(2026, 9, 11));
        rr.DisplayReview();

    
```
