```polyglot-notebook
using System;
using System.Collections.Generic;
using System.Linq;

public interface IInstallable
{
    void Install();
}

public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    protected Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}, Цена: {Price} ₽");
    }

    public override string ToString()
    {
        return $"{GetType().Name} '{Name}'  {Price} ₽";
    }
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public double FrequencyGHz { get; set; }

    public CPU(string name, decimal price, int cores, double frequencyGHz)
        : base(name, price)
    {
        Cores = cores;
        FrequencyGHz = frequencyGHz;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  Ядер: {Cores}, Частота: {FrequencyGHz} GHz");
    }

    public void Install()
    {
        Console.WriteLine($"Установлен процессор: {Name}");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(CPU a, CPU b) => !(a == b);

    public override bool Equals(object obj) => obj is CPU other && this == other;
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public class RAM : Component, IInstallable
{
    public int CapacityGB { get; set; }
    public int SpeedMHz { get; set; }

    public RAM(string name, decimal price, int capacityGB, int speedMHz)
        : base(name, price)
    {
        CapacityGB = capacityGB;
        SpeedMHz = speedMHz;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  Объём: {CapacityGB} GB, Частота: {SpeedMHz} MHz");
    }

    public void Install()
    {
        Console.WriteLine($"Установлена память: {Name}");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(RAM a, RAM b) => !(a == b);

    public override bool Equals(object obj) => obj is RAM other && this == other;
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public class GPU : Component, IInstallable
{
    public int MemoryGB { get; set; }
    public string Manufacturer { get; set; }

    public GPU(string name, decimal price, int memoryGB, string manufacturer)
        : base(name, price)
    {
        MemoryGB = memoryGB;
        Manufacturer = manufacturer;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  Память: {MemoryGB} GB, Производитель: {Manufacturer}");
    }

    public void Install()
    {
        Console.WriteLine($"Установлена видеокарта: {Name}");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(GPU a, GPU b) => !(a == b);

    public override bool Equals(object obj) => obj is GPU other && this == other;
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public class HDD : Component, IInstallable
{
    public int CapacityGB { get; set; }
    public int SpeedRPM { get; set; }

    public HDD(string name, decimal price, int capacityGB, int speedRPM)
        : base(name, price)
    {
        CapacityGB = capacityGB;
        SpeedRPM = speedRPM;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  Объём: {CapacityGB} GB, Скорость: {SpeedRPM} RPM");
    }

    public void Install()
    {
        Console.WriteLine($"Установлен диск: {Name}");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(HDD a, HDD b) => !(a == b);

    public override bool Equals(object obj) => obj is HDD other && this == other;
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}


public class Computer
{
    private List<Component> _components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (_components.Remove(component))
            Console.WriteLine($"Удалён компонент: {component.Name}");
        else
            Console.WriteLine($"Компонент {component.Name} не найден.");
    }

    public decimal GetTotalPrice()
    {
        return _components.Sum(c => c.Price);
    }

    public void ShowAllComponents()
    {
        Console.WriteLine(" Состав компьютера ");
        foreach (var c in _components)
        {
            c.GetInfo();
            Console.WriteLine();
        }
    }
}


public static class ComputerFactory
{
    public static Computer CreateOfficePC()
    {
        Computer pc = new Computer();
        pc.AddComponent(new CPU("Intel Core i3", 10000m, 2, 3.5));
        pc.AddComponent(new RAM("Kingston 8GB", 3000m, 8, 2400));
        pc.AddComponent(new HDD("Seagate 500GB", 2500m, 500, 7200));
        return pc;
    }

    public static Computer CreateGamingPC()
    {
        Computer pc = new Computer();
        pc.AddComponent(new CPU("Intel Core i9", 50000m, 16, 5.0));
        pc.AddComponent(new RAM("Corsair 32GB", 12000m, 32, 3200));
        pc.AddComponent(new GPU("NVIDIA RTX 4090", 150000m, 24, "NVIDIA"));
        pc.AddComponent(new HDD("Samsung SSD 2TB", 15000m, 2000, 7200));
        return pc;
    }

    public static Computer CreateCustomPC<T>(T component) where T : Component, IInstallable
    {
        Computer pc = new Computer();
        pc.AddComponent(component);
        return pc;
    }
}

```

```polyglot-notebook
Computer office = ComputerFactory.CreateOfficePC();
office.ShowAllComponents();
```

```polyglot-notebook
Computer gaming = ComputerFactory.CreateGamingPC();
gaming.ShowAllComponents();
```

```polyglot-notebook
Computer custom = ComputerFactory.CreateCustomPC(new GPU("RTX 3080", 80000m, 10, "NVIDIA"));
custom.ShowAllComponents();
```
